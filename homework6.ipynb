```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
```

```python
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
y = df.fuel_efficiency_mpg
X = df.drop('fuel_efficiency_mpg', axis=1)
```

```python
df_full_train, df_test = train_test_split(
    df, test_size=0.2, random_state=1
)
```

```python
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1
)
```

```python
len(df_train), len(df_val), len(df_test)
```

```python
df_train = df_train.fillna(0)
df_val = df_val.fillna(0)
df_test = df_test.fillna(0)
```

```python
df_train.isnull().sum()
```

```python
y_train = df_train.fuel_efficiency_mpg.values
y_val = df_val.fuel_efficiency_mpg.values
y_test = df_test.fuel_efficiency_mpg.values

del df_train['fuel_efficiency_mpg']
del df_val['fuel_efficiency_mpg']
del df_test['fuel_efficiency_mpg']

X_train = df_train
X_val = df_val
X_test = df_test
```

```python
train_dicts = X_train.to_dict(orient='records')
val_dicts = X_val.to_dict(orient='records')
test_dicts = X_test.to_dict(orient='records')
# dictvectorizor need directory 
```

```python
# fit learns the feature names and categorical values from training data.
# transform applies that same learned structure to validation and test data.
dv = DictVectorizer(sparse=True)

X_train = dv.fit_transform(train_dicts)
X_val = dv.transform(val_dicts)
X_test = dv.transform(test_dicts)
```

```python
X_train.shape, X_val.shape, X_test.shape
```

```python
from sklearn.tree import DecisionTreeRegressor, export_text

dt = DecisionTreeRegressor(max_depth=1)
dt.fit(X_train, y_train)
```

```python
print(export_text(dt, feature_names=list(dv.get_feature_names_out())))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=10,
    random_state=1,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_val)
```

```python
rmse = mean_squared_error(y_val, y_pred) ** 0.5

round(rmse, 4)
```

```python
scores = []

for n in [10, 50, 100, 150]:
    rf = RandomForestRegressor(
        n_estimators=n,
        random_state=1,
        n_jobs=-1
    )

    rf.fit(X_train, y_train)

    y_pred = rf.predict(X_val)

    rmse = mean_squared_error(y_val, y_pred) ** 0.5

    scores.append((n, rmse))

scores
```

```python
for depth in [10, 15, 20, 25]:
    scores = []

    for n in [10, 50, 100, 150]:
        rf = RandomForestRegressor(
            n_estimators=n,
            max_depth=depth,
            random_state=1,
            n_jobs=-1
        )

        rf.fit(X_train, y_train)
        y_pred = rf.predict(X_val)

        rmse = mean_squared_error(y_val, y_pred) ** 0.5
        scores.append(rmse)

    print(depth, round(np.mean(scores), 4))
```

```python
rf = RandomForestRegressor(
    n_estimators=10,
    max_depth=20,
    random_state=1,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
importances = rf.feature_importances_

feature_names = dv.get_feature_names_out()

list(zip(feature_names, importances))
```

```python
for feature, importance in zip(feature_names, importances):
    if feature in [
        'vehicle_weight',
        'horsepower',
        'acceleration',
        'engine_displacement'
    ]:
        print(feature, importance)
```

```python
import xgboost as xgb
```

```python
features = list(dv.get_feature_names_out())

dtrain = xgb.DMatrix(
    X_train,
    label=y_train,
    feature_names=features
)

dval = xgb.DMatrix(
    X_val,
    label=y_val,
    feature_names=features
)
```

```python
xgb_params = {
    'eta': 0.3,
    'max_depth': 6,
    'min_child_weight': 1,
    'objective': 'reg:squarederror',
    'nthread': 8,
    'seed': 1,
    'verbosity': 1
}
```

```python
model = xgb.train(
    xgb_params,
    dtrain,
    num_boost_round=100
)
```

```python
y_pred = model.predict(dval)
```

```python
rmse = mean_squared_error(y_val, y_pred) ** 0.5

round(rmse, 4)
```

```python
xgb_params['eta'] = 0.1
```

```python
model_01 = xgb.train(
    xgb_params,
    dtrain,
    num_boost_round=100
)
```

```python
y_pred_01 = model_01.predict(dval)

rmse_01 = mean_squared_error(y_val, y_pred_01) ** 0.5

round(rmse_01, 4)
```
